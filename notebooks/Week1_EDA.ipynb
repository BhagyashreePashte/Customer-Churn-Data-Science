```python
before_rows = len(df_clean)

df_clean = df_clean.dropna(subset=["TotalCharges"])

after_rows = len(df_clean)

print("Rows before cleaning:", before_rows)
print("Rows after cleaning:", after_rows)
print("Rows removed:", before_rows - after_rows)
```

```python
categorical_columns = df_clean.select_dtypes(
    include="object"
).columns

for column in categorical_columns:
    print(
        column,
        "→",
        df_clean[column].astype(str).str.strip().unique()[:10]
    )
```

```python
for column in categorical_columns:
    df_clean[column] = df_clean[column].str.strip()
```

```python
print("Whitespace cleaning completed.")
```

```python
print(
    "Duplicate rows after cleaning:",
    df_clean.duplicated().sum()
)

print(
    "Duplicate customer IDs after cleaning:",
    df_clean["customerID"].duplicated().sum()
)
```

```python
print("========== FINAL DATA VALIDATION ==========")

print("Rows:", df_clean.shape[0])
print("Columns:", df_clean.shape[1])

print(
    "Missing values:",
    df_clean.isnull().sum().sum()
)

print(
    "Duplicate rows:",
    df_clean.duplicated().sum()
)

print(
    "Duplicate customer IDs:",
    df_clean["customerID"].duplicated().sum()
)

print(
    "TotalCharges datatype:",
    df_clean["TotalCharges"].dtype
)

print("===========================================")
```

```python
processed_path = "../data/processed/Telco-Customer-Churn-Cleaned.csv"

df_clean.to_csv(
    processed_path,
    index=False
)

print("Cleaned dataset saved successfully!")
print(processed_path)
```

```python
# Select only the numerical columns we need
numeric_columns = [
    "SeniorCitizen",
    "tenure",
    "MonthlyCharges",
    "TotalCharges"
]

# Convert numerical columns to numeric values
for column in numeric_columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")

# Check missing values created during conversion
print("Missing values after numeric conversion:")
print(df[numeric_columns].isnull().sum())

# Calculate correlation
correlation_matrix = df[numeric_columns].corr()

correlation_matrix
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix of Numerical Variables")

plt.tight_layout()

plt.savefig(
    "../visualizations/09_correlation_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="tenure",
    y="MonthlyCharges",
    hue="Churn",
    alpha=0.6
)

plt.title("Tenure vs Monthly Charges by Churn Status")
plt.xlabel("Tenure (Months)")
plt.ylabel("Monthly Charges")

plt.tight_layout()

plt.savefig(
    "../visualizations/10_tenure_vs_monthly_charges.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
